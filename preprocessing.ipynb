```python
plt.figure(figsize = (12,5))

plt.plot(df["Timestamp"],df["PM2.5 (µg/m³)"])

plt.xlabel("Time")
plt.ylabel("PM2.5 (µg/m³)")
plt.title("PM2.5 Concentration Over Time")

plt.show()
```

```python
# to understand seasonal patterns ------ > monthly average of PM2.5

df["Month"] = df["Timestamp"].dt.month      # created new column
monthly_PM25 = df.groupby("Month")["PM2.5 (µg/m³)"].mean()
monthly_PM25
```

```python
# visually understand seasonal patterns

plt.figure(figsize=(10,5))

plt.plot(monthly_PM25.index,monthly_PM25.values,marker="o")

plt.xlabel("Month")
plt.ylabel("Average PM2.5 (µg/m³)")
plt.title("Monthly Average PM2.5")

plt.xticks(range(1,13))

plt.show()
```

```python
# correlation analysis of experiment#1

cams_composition = ['co', 'aermr04', 'aermr05', 'aermr06', 'aermr09', 'aermr07', 'aermr10',
       'aermr08', 'no2', 'no', 'go3','so2']

corr_df = df[["PM2.5 (µg/m³)"] + cams_composition]
```

```python
corr_matrix = corr_df.corr()
corr_matrix
```

```python
# let's visually understand through heatmap

plt.figure(figsize = (8,6))

plt.imshow(corr_matrix,cmap = "coolwarm",aspect= "auto")

plt.colorbar(label = "Correlation")

plt.xticks(range(len(corr_matrix.columns)),corr_matrix.columns,rotation=90)
plt.yticks(range(len(corr_matrix.columns)),corr_matrix.columns)

plt.title("Correlation Heatmap - CAMS composition and CPCB PM2.5")

plt.tight_layout()
plt.show()
```

```python
cams_meteorology = ['q','t','u10', 'v10', 'd2m', 't2m']

exp2_df = df[["PM2.5 (µg/m³)"] + cams_composition + cams_meteorology]

exp2_corr = exp2_df.corr()
exp2_corr
```

```python
plt.figure(figsize = (8,6))

plt.imshow(exp2_corr,cmap = "coolwarm",aspect= "auto")

plt.colorbar(label = "Correlation")

plt.xticks(range(len(exp2_corr.columns)),exp2_corr.columns,rotation=90)
plt.yticks(range(len(exp2_corr.columns)),exp2_corr.columns)

plt.title("Correlation Heatmap - CAMS Atmospheric Composition, Meteorological Parameters and CPCB PM2.5")

plt.tight_layout()
plt.show()
```

```python
# multicollinearity of experiment 1 

X1 = df[cams_composition]

```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif1 = pd.DataFrame()

vif1["Feature"] = X1.columns

vif1["VIF"] = [
    variance_inflation_factor(X1.values, i)
    for i in range(X1.shape[1])
]

vif1
```

```python
# to find out multicollinearity of experiment2


X = df[cams_composition + cams_meteorology]

vif = pd.DataFrame()

vif["Feature"] = X.columns

vif["VIF"] = [
    variance_inflation_factor(X.values, i)
    for i in range(X.shape[1])
]

vif
```

```python
# providing folder location

from pathlib import Path

cams_folder = Path(r"C:\Users\admin\Documents\My_projects\AQI_project\data\raw\CAMS\Multi_level_model60")
```

```python
print(cams_folder)
print(list(cams_folder.glob("*.grib")))
```

```python
# Process all GRIB files automatically

cams_dfs = []

for file_path in cams_folder.glob("*.grib"):
    print("processing : ", file_path.name)

    ds = xr.open_dataset(
        file_path,
        engine = "cfgrib"
    )

# select nearest CAMS grid point

    g_point = ds.sel(
        latitude = 28.64,
        longitude = 77.31,
        method= "nearest"
    )

# convert to dataframe

    df = g_point.to_dataframe().reset_index()

# store it 

    cams_dfs.append(df)

    ds.close()

print("Number of files processed :", len(cams_dfs))
```

```python
# combine them

cams_combined = pd.concat(
    cams_dfs,
    ignore_index=True 
)

print(cams_combined.shape)
print(cams_combined.head())
```

```python
cams_combined = cams_combined.sort_values("time").reset_index(drop = True)
```

```python
output_path = r"D:\AQI_project\CAMS\cams_multi_level_integrated.csv"
cams_combined.to_csv(output_path,index=False)
```

```python
# single level data integration
import xarray as xr
import pandas as pd
from pathlib import Path
cams_single_folder = Path(r"C:\Users\admin\Documents\My_projects\AQI_project\data\raw\CAMS\single_level")
```

```python
c_dfs = []

for file_path in cams_single_folder.glob("*.grib"):
    print("processing : ", file_path.name)
    
    ds = xr.open_dataset(
        file_path,
        engine = "cfgrib"
    )
    
# select nearest CAMS grid point
    
    grid_point = ds.sel(
        latitude = 28.64,
        longitude = 77.31,
        method= "nearest"
    )
    
# convert to dataframe
    
    s_df = grid_point.to_dataframe().reset_index()
    
# store it 
    
    c_dfs.append(s_df)
    
    ds.close()
    
print("Number of files processed :", len(c_dfs))
```

```python
cams_single_combined = pd.concat(
    c_dfs,
    ignore_index=True 
)

print(cams_single_combined.shape)
print(cams_single_combined.head())
```

```python
cams_single_combined = cams_single_combined.sort_values("time").reset_index(drop = True)
```

```python
output_path = r"D:\AQI_project\CAMS\cams_single_level_integrated.csv"
cams_single_combined.to_csv(output_path,index=False)
```
